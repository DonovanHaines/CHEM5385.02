```python
# 1. Inputs and target output
X = [0.5, -0.2]   # Features
Y_true = 1.0      # What the network should predict
                  # (Given those x-inputs, the right answer is 1.0)

# 2. Initialization
W = [0.1, -0.1]   # Weights
b = 0.0           # Bias starts at 0 (not 1)

print(f"Initial Bias: {b}")

# 3. Forward pass (Calculate prediction)
# Output = (x1 * w1) + (x2 * w2) + bias
z = (X[0] * W[0]) + (X[1] * W[1]) + b
Y_pred = z
print(f"Prediction: {Y_pred}") # Output is 0.07, which is far from 1.0

# 4. Backpropagation (Calculate errors and gradients)
# How far off were we?
error = Y_pred - Y_true
loss_grad = 2 * error

# The gradient for a bias is always just the loss gradient
# because d(wx + b)/db = 1
db = loss_grad

# 5. Gradient Descent (Update parameters)
lr = 0.5  # Learning rate
b -= lr * db

print(f"Updated Bias after 1 step: {b:.4f}")

# 6. Update Calculate prediction
# Output = (x1 * w1) + (x2 * w2) + bias
z = (X[0] * W[0]) + (X[1] * W[1]) + b
Y_pred = z
print(f"Prediction: {Y_pred}") # Output is 0.07, which is far from 1.0
```
